```python
# Assuming 'word' is already defined from the previous input cell
# If not, you can uncomment the line below:
# word = str(input("Enter a word: "))

uppercase_word = ""
for char in word:
    if 'a' <= char <= 'z':
        # Convert lowercase to uppercase using ASCII values
        # The difference between 'a' and 'A' is 32
        uppercase_word += chr(ord(char) - 32)
    else:
        uppercase_word += char

print(f"Original word: {word}")
print(f"Uppercase word: {uppercase_word}")
```
