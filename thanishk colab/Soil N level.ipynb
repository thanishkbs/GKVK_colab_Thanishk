```python
Crop=str(input("crop:"))
Soil_moisture=int(input("soil moisture:"))
Temperature=str(input("Temparature:"))
print(f" Crop: {Crop}, Soil moisture: {Soil_moisture}%, Temperature:{Temperature}")

if Crop=="rice":
  for rice in Crop:
    if Soil_moisture<60 and Temperature=="hot":
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(60%) for rice\n estimated water needed=110mm")
    elif Soil_moisture<50 and (Temperature=="moderate" or Temperature=="cool"):
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(50%) for rice\nEstimated water needed=110mm")
    else:
      print("Do not irrigate")
    break
elif Crop== "wheat":
  for wheat in Crop:
    if Soil_moisture<40 and Temperature=="hot":
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(60%) for wheat\n estimated water needed=80mm")
    elif Soil_moisture<30 and (Temperature=="moderate" or Temperature=="cool"):
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(50%) for wheat\nEstimated water needed=80mm")
    else:
      print("Do not irrigate")
    break
elif Crop== "sugarcane":
  for sugarcane in Crop:
    if Soil_moisture<70 and Temperature=="hot":
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(60%) for Sugarcane\n estimated water needed=200mm")
    elif Soil_moisture<60 and (Temperature=="moderate" or Temperature=="cool"):
      print(f"Decision: Irrigate NOW\nReason: Soil moisture is{Soil_moisture},it is below threshold soil moisture(50%) for Sugarcane\nEstimated water needed=200mm")
    else:
      print("Do not irrigate")
    break
else:
  print("Do not irrigate")
```
