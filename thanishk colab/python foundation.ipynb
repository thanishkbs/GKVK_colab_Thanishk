```python
marks=[56,78,90,45,88]
#max
max_mark=max(marks)

#average
total=0
for m in marks:
 total=sum(marks)
average=total/len(marks)

#count
count=0
for m in marks:
  if m>=60:
    count=count+1

print(f"Max={max_mark}")
print(f"Total={total}")
print(f"Average={average}")
print(f"Number of marks >= 60: {count}")
```

```python
def average_marks(marks_list):
  total=0
  for m in marks_list:
   total=total+m
  avg=total/len(marks_list)
  return avg
data=[100,90,50]
print(f"Average=",average_marks(data))
```
